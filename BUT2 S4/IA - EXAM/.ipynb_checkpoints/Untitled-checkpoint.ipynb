### Exercice 1

```python
# Import bibliotheque
import numpy as np

#Question 1. 
def affine_generale(b, w, x):
    return np.dot(w, x) + b

#Question 2.
def activation(s):
    if s >= 0 :
        return 1 
    else :
        return 0

#Question 3.
def perceptron(b, w, x):
    s = affine_generale(b, w, x)
    return activation(s)

#Partie de test 

b_test = 1
w_test = np.array([2, -1])

vec_entree = [np.array([-2, 1]), np.array([0, 2]), np.array([0, 0]), np.array([2, 1])]

print("Test de l'exo 1 :")
for x in vec_entree:
    res = perceptron(b_test, w_test, x)
    print(f"Entrée x={x} -> Sortie : {res}")
```
