```python
#import des bibliotheques
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

iris = load_iris()
data = iris.data

target = np.concatenate((np.zeros(50, dtype=int), np.ones(50, dtype=int), np.zeros(50, dtype=int)))

np.random.seed(10)
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.4)
cmp = np.array(['r', 'g'])
```

```python
#visualisation en 2d
plt.figure(figsize=(6, 4))
plt.scatter(data[:, 2], data[:, 3], c=cmp[target], s=50, edgecolors='none')
plt.title("Visualisation 2D")
plt.xlabel("Longueur pétale (cm)")
plt.ylabel("Largeur pétale (cm)")
plt.show()
```

### Exercice 1

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(data[:, 1], data[:, 2], data[:, 3], c=cmp[target], s=50)

ax.set_title("visualisation en 3D")
ax.set_xlabel("largeur sépale")
ax.set_ylabel("longueur pétale")
ax.set_zlabel("largeur pétale")
plt.show()
```

```python
np.random.seed(10) # initialiser le generateur 
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
X_train1, X_test1, y_train1, y_test1 = train_test_split(data, target, test_size=0.5)

clf = MLPClassifier(solver='sgd', learning_rate_init=0.01, max_iter=10000, random_state=0) 

# apprentissage avec les deux premieres variables 
clf.fit(X_train1[:,2:4], y_train1) 

train_score = clf.score(X_train1[:,2:4], y_train1) 
print("Le score en train est {}".format(train_score))

test_score = clf.score(X_test1[:,2:4], y_test1)
print("Le score en test est {}".format(test_score))
```

```python
# creer une nouvelle figure
plt.figure() 

# afficher les nuages de points d'apprentissage et de test
plt.scatter(X_train1[:,2], X_train1[:,3], c=cmp[y_train1], s=50, edgecolors='none', label='Train') 
plt.scatter(X_test1[:,2], X_test1[:,3], c='none', s=50, edgecolors=cmp[y_test1], label='Test')

# calculer la probabilite de sortie du MLP 
nx, ny = 200, 200
x_min, x_max = plt.xlim()
y_min, y_max = plt.ylim() 
xx, yy = np.meshgrid(np.linspace(x_min, x_max, nx), np.linspace(y_min, y_max, ny))

# Prediction sur la grille
Z = clf.predict_proba(np.c_[xx.ravel(), yy.ravel()])
Z1 = Z[:, 1].reshape(xx.shape) 

# dessiner le contour
plt.contour(xx, yy, Z1, [0.5], colors='black')

plt.title("Frontière de séparation")
plt.xlabel("Longueur pétale")
plt.ylabel("Largeur pétale")
plt.legend()
plt.show()
```

```python
from matplotlib import cm
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Nuages de points
ax.scatter(X_train1[:, 2], X_train1[:, 3], 0.5, c=cmp[y_train1], s=50)
ax.scatter(X_test1[:, 2], X_test1[:, 3], 0.5, c=cmp[y_test1], s=50)

# Limites
x_min, x_max = plt.xlim()
y_min, y_max = plt.ylim()

# Création de la grille
xx, yy = np.meshgrid(np.linspace(x_min, x_max, nx), np.linspace(y_min, y_max, ny))

# Calculs
Z = clf.predict_proba(np.c_[xx.ravel(), yy.ravel()])
Z0 = Z[:, 0].reshape(xx.shape)
Z1 = Z[:, 1].reshape(xx.shape)

# Surfaces
# probabilite d'appartenance a la classe Versicolor
surf1 = ax.plot_surface(xx, yy, Z0, alpha=0.5, cmap=cm.Reds)
# probabilite d'appartenance a une autre classe
surf2 = ax.plot_surface(xx, yy, Z1, alpha=0.5, cmap=cm.Greens)

# Barres de couleur
fig.colorbar(surf1, shrink=0.5, aspect=5)
fig.colorbar(surf2, shrink=0.5, aspect=5)

plt.show()
```

### Exercice 5

```python
train_scores = []
test_scores = []
for hlsize in [2, 10, 50, 100, 150, 200]:
    clf = MLPClassifier(hidden_layer_sizes=(hlsize,), max_iter=3000, random_state=0)
    clf.fit(X_train1[:,2:4], y_train1)
    train_scores.append(clf.score(X_train1[:,2:4], y_train1))
    test_scores.append(clf.score(X_test1[:,2:4], y_test1))

plt.figure()
plt.xlabel('neurones caches')
plt.ylabel('scores')
plt.plot([2, 10, 50, 100, 150, 200], train_scores, 'go')
plt.plot([2, 10, 50, 100, 150, 200], test_scores, 'r+')
plt.show()
```

Il n'est pas necessaire d'ajouter trop de neurones, ca enleve trop de précision et ça charge le programme inutilement

```python
train_scores = []
test_scores = []
nb_couches_liste = [2, 3, 4, 5]

for n in nb_couches_liste:
    layers = tuple([10] * n)
    
    clf = MLPClassifier(hidden_layer_sizes=layers, activation='relu', max_iter=5000, random_state=0)
    
    clf.fit(X_train1[:, 2:4], y_train1)
    
    train_scores.append(clf.score(X_train1[:, 2:4], y_train1))
    test_scores.append(clf.score(X_test1[:, 2:4], y_test1))

plt.figure()
plt.xlabel('Nombre de couches cachées')
plt.ylabel('Scores')
plt.plot(nb_couches_liste, train_scores, 'go-', label='Train')
plt.plot(nb_couches_liste, test_scores, 'r+-', label='Test')
plt.legend()
plt.title("Impact du nombre de couches (10 neurones/couche)")
plt.show()
```

on constate que les scores stagnent, ajouter des couches cachées n'améliore pas les performances . c'est inutile car le problème est trop simple pour nécessiter un tel réseau
