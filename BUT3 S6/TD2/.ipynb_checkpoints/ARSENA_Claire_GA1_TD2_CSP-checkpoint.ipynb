### Exercice 1

À partir du formalisme introduit en cours ($P = (X, D, C, R)$), on modélise le Sudoku comme un CSP :

- **X** (variables) : une variable $x_{ij}$ pour chaque case de la grille, avec $i, j \in \{0,...,8\}$, soit 81 variables.
- **D** (domaines) : pour chaque case déjà remplie, $d_{ij} = \{v\}$ (domaine réduit à la valeur imposée) ; pour chaque case vide, $d_{ij} = \{1,...,9\}$.
- **C** (contraintes) : 
  - pour chaque ligne $i$, une contrainte "all-different" portant sur $\{x_{i0}, x_{i1}, ..., x_{i8}\}$
  - pour chaque colonne $j$, une contrainte "all-different" portant sur $\{x_{0j}, x_{1j}, ..., x_{8j}\}$
  - pour chaque bloc $3\times3$, une contrainte "all-different" portant sur les 9 cases du bloc
- **R** (relations) : pour chaque contrainte $c$ (portant sur 9 variables), $r_c$ = ensemble des 9-uplets dont toutes les valeurs sont distinctes deux à deux, c'est-à-dire les permutations de $\{1,...,9\}$.

Une affectation complète est une solution si et seulement si elle satisfait toutes les contraintes all-different, ce qui revient à ce qu'aucune ligne, colonne ou bloc ne contienne deux fois la même valeur.

### Exercice 2

On représente la grille par une liste de listes (9x9), où 0 désigne une case vide. On écrit la fonction `Backtrack` qui explore récursivement l'arbre d'affectation en suivant l'algorithme du cours : on choisit une case vide, on essaie chaque valeur de son domaine, on vérifie la consistance (ligne/colonne/bloc), et on backtrack en cas d'échec.

```python
import copy
import time

# grille du TD (0 = case vide)
grille_initiale = [
    [4, 0, 0, 3, 0, 0, 7, 6, 0],
    [0, 9, 3, 0, 7, 0, 4, 0, 1],
    [2, 0, 0, 1, 4, 0, 0, 8, 3],
    [9, 5, 8, 0, 0, 0, 0, 0, 7],
    [3, 4, 0, 0, 0, 7, 0, 1, 0],
    [0, 7, 2, 8, 9, 3, 5, 4, 0],
    [0, 8, 0, 2, 0, 0, 3, 7, 4],
    [0, 0, 4, 0, 0, 0, 1, 9, 5],
    [0, 0, 0, 0, 5, 0, 6, 0, 0],
]

def afficher_grille(g):
    for i, row in enumerate(g):
        if i % 3 == 0 and i != 0:
            print("-" * 21)
        ligne = ""
        for j, val in enumerate(row):
            if j % 3 == 0 and j != 0:
                ligne += "| "
            ligne += (str(val) if val != 0 else ".") + " "
        print(ligne)

print("Grille initiale :")
afficher_grille(grille_initiale)
```

```python
def est_consistant(grille, ligne, col, val):
    """Vérifie que la valeur val peut être placée en (ligne, col)
    sans violer les contraintes de ligne, colonne et bloc 3x3."""
    # contrainte de ligne
    for j in range(9):
        if grille[ligne][j] == val:
            return False
    # contrainte de colonne
    for i in range(9):
        if grille[i][col] == val:
            return False
    # contrainte de bloc 3x3
    bl, bc = 3 * (ligne // 3), 3 * (col // 3)
    for i in range(bl, bl + 3):
        for j in range(bc, bc + 3):
            if grille[i][j] == val:
                return False
    return True

def case_vide(grille):
    """Retourne les coordonnées (ligne, col) de la première case vide,
    ou None si la grille est complète."""
    for i in range(9):
        for j in range(9):
            if grille[i][j] == 0:
                return (i, j)
    return None

def Backtrack(grille):
    """Algorithme de backtracking simple pour résoudre la grille (en place).
    Retourne True si une solution a été trouvée."""
    case = case_vide(grille)
    if case is None:
        return True  # V est vide : A est une solution

    ligne, col = case
    for v in range(1, 10):  # domaine dx = {1, ..., 9}
        if est_consistant(grille, ligne, col, v):
            grille[ligne][col] = v
            if Backtrack(grille):
                return True
            grille[ligne][col] = 0  # retour en arrière (backtrack)
    return False
```

```python
grille_bt = copy.deepcopy(grille_initiale)

debut = time.time()
solution_trouvee = Backtrack(grille_bt)
duree = time.time() - debut

print("Solution trouvée :", solution_trouvee)
print("Temps de résolution : {:.4f} s".format(duree))
print()
afficher_grille(grille_bt)
```

Le backtracking simple résout la grille quasi instantanément. Chaque fois qu'une valeur viole une contrainte (ligne, colonne ou bloc déjà utilisé), on passe directement à la valeur suivante sans explorer plus loin la branche, ce qui évite de parcourir tout l'espace des affectations complètes (contrairement au Generate and Test vu en cours).

### Exercice 3

On intègre le Forward Checking : à chaque affectation d'une valeur, on retire cette valeur des domaines des variables voisines (même ligne, même colonne, même bloc). Si un domaine devient vide, on détecte l'échec immédiatement, sans attendre d'instancier la variable concernée.

```python
def voisins(ligne, col):
    """Retourne l'ensemble des cases liées par une contrainte à (ligne, col) :
    même ligne, même colonne, même bloc 3x3."""
    v = set()
    for j in range(9):
        if j != col:
            v.add((ligne, j))
    for i in range(9):
        if i != ligne:
            v.add((i, col))
    bl, bc = 3 * (ligne // 3), 3 * (col // 3)
    for i in range(bl, bl + 3):
        for j in range(bc, bc + 3):
            if (i, j) != (ligne, col):
                v.add((i, j))
    return v

def init_domaines(grille):
    """Construit le dictionnaire des domaines dx pour chaque case,
    en tenant compte des valeurs déjà présentes dans la grille."""
    domaines = {}
    for i in range(9):
        for j in range(9):
            if grille[i][j] == 0:
                domaines[(i, j)] = set(range(1, 10))
            else:
                domaines[(i, j)] = {grille[i][j]}
    for i in range(9):
        for j in range(9):
            if grille[i][j] != 0:
                for (pi, pj) in voisins(i, j):
                    domaines[(pi, pj)].discard(grille[i][j])
    return domaines
```

```python
def Forward_Checking(grille, domaines, cases_libres):
    """Algorithme de Forward Checking : à chaque affectation, on met à jour
    (et on restaure si besoin) les domaines des cases voisines."""
    if not cases_libres:
        return True  # V est vide : A est une solution

    x = cases_libres[0]  # ordre naturel (pas d'heuristique dans cet exercice)
    ligne, col = x
    reste = cases_libres[1:]

    for v in sorted(domaines[x]):  # d <- dx
        # Sauvegarde et mise à jour des domaines voisins (Check-Forward)
        sauvegarde = []
        consistant = True
        grille[ligne][col] = v

        for xj in voisins(ligne, col):
            if xj in domaines and v in domaines[xj]:
                domaines[xj].discard(v)
                sauvegarde.append(xj)
                if len(domaines[xj]) == 0 and grille[xj[0]][xj[1]] == 0:
                    consistant = False

        if consistant and Forward_Checking(grille, domaines, reste):
            return True

        # Restauration
        for xj in sauvegarde:
            domaines[xj].add(v)
        grille[ligne][col] = 0

    return False
```

```python
grille_fc = copy.deepcopy(grille_initiale)
domaines_fc = init_domaines(grille_fc)
cases_libres = [(i, j) for i in range(9) for j in range(9) if grille_fc[i][j] == 0]

debut = time.time()
solution_trouvee = Forward_Checking(grille_fc, domaines_fc, cases_libres)
duree = time.time() - debut

print("Solution trouvée :", solution_trouvee)
print("Temps de résolution : {:.4f} s".format(duree))
print()
afficher_grille(grille_fc)
```

On constate que le Forward Checking résout la grille en explorant moins de nœuds que le backtracking simple : les domaines contraints et les domaines vides sont détectés dès la mise à jour, ce qui permet de couper des branches plus tôt dans l'arbre de recherche, avant même d'essayer d'instancier la variable concernée.

### Exercice 4

L'ordre d'instanciation des **valeurs** n'a aucune influence sur les problèmes inconsistants, car un problème inconsistant est un problème qui ne possède aucune solution : quel que soit l'ordre dans lequel on essaie les valeurs d'un domaine, l'algorithme finira par toutes les essayer (et échouer) avant de remonter, puisqu'aucune affectation complète n'est consistante. Changer l'ordre ne fait donc que changer l'ordre dans lequel les branches (toutes vouées à l'échec) sont explorées, mais pas le fait qu'elles échouent toutes : l'espace de recherche parcouru reste, au pire, le même arbre entier. Seul l'ordre d'instanciation des **variables** peut avoir un effet, en permettant de détecter l'échec plus tôt (principe de l'échec d'abord).

### Exercice 5

On modifie le Forward Checking pour qu'il choisisse, à chaque étape, la variable dont le domaine courant est le plus petit (heuristique MRV : Minimum Remaining Values), plutôt que de suivre l'ordre naturel des cases.

```python
def Forward_Checking_MRV(grille, domaines, cases_libres):
    """Forward Checking avec heuristique dynamique de choix de variable :
    on instancie en priorité la variable ayant le plus petit domaine."""
    if not cases_libres:
        return True

    # heuristique : variable de domaine minimal
    x = min(cases_libres, key=lambda c: len(domaines[c]))
    ligne, col = x
    reste = [c for c in cases_libres if c != x]

    for v in sorted(domaines[x]):
        sauvegarde = []
        consistant = True
        grille[ligne][col] = v

        for xj in voisins(ligne, col):
            if xj in domaines and v in domaines[xj]:
                domaines[xj].discard(v)
                sauvegarde.append(xj)
                if len(domaines[xj]) == 0 and grille[xj[0]][xj[1]] == 0:
                    consistant = False

        if consistant and Forward_Checking_MRV(grille, domaines, reste):
            return True

        for xj in sauvegarde:
            domaines[xj].add(v)
        grille[ligne][col] = 0

    return False
```

```python
grille_mrv = copy.deepcopy(grille_initiale)
domaines_mrv = init_domaines(grille_mrv)
cases_libres_mrv = [(i, j) for i in range(9) for j in range(9) if grille_mrv[i][j] == 0]

debut = time.time()
solution_trouvee = Forward_Checking_MRV(grille_mrv, domaines_mrv, cases_libres_mrv)
duree = time.time() - debut

print("Solution trouvée :", solution_trouvee)
print("Temps de résolution : {:.4f} s".format(duree))
print()
afficher_grille(grille_mrv)
```

```python
# Comparaison du nombre de nœuds explorés par les 3 approches

def compter_noeuds(fonction_resolution, *args):
    compteur = [0]
    fonction_orig = fonction_resolution
    return compteur

# On instrumente chaque algorithme avec un compteur global de nœuds explorés
def Backtrack_compte(grille, compteur):
    compteur[0] += 1
    case = case_vide(grille)
    if case is None:
        return True
    ligne, col = case
    for v in range(1, 10):
        if est_consistant(grille, ligne, col, v):
            grille[ligne][col] = v
            if Backtrack_compte(grille, compteur):
                return True
            grille[ligne][col] = 0
    return False

def Forward_Checking_compte(grille, domaines, cases_libres, compteur, mrv=False):
    compteur[0] += 1
    if not cases_libres:
        return True
    if mrv:
        x = min(cases_libres, key=lambda c: len(domaines[c]))
        reste = [c for c in cases_libres if c != x]
    else:
        x = cases_libres[0]
        reste = cases_libres[1:]
    ligne, col = x
    for v in sorted(domaines[x]):
        sauvegarde = []
        consistant = True
        grille[ligne][col] = v
        for xj in voisins(ligne, col):
            if xj in domaines and v in domaines[xj]:
                domaines[xj].discard(v)
                sauvegarde.append(xj)
                if len(domaines[xj]) == 0 and grille[xj[0]][xj[1]] == 0:
                    consistant = False
        if consistant and Forward_Checking_compte(grille, domaines, reste, compteur, mrv):
            return True
        for xj in sauvegarde:
            domaines[xj].add(v)
        grille[ligne][col] = 0
    return False

# Backtrack simple
g1 = copy.deepcopy(grille_initiale)
c1 = [0]
Backtrack_compte(g1, c1)

# Forward checking (ordre naturel)
g2 = copy.deepcopy(grille_initiale)
d2 = init_domaines(g2)
cl2 = [(i, j) for i in range(9) for j in range(9) if g2[i][j] == 0]
c2 = [0]
Forward_Checking_compte(g2, d2, cl2, c2, mrv=False)

# Forward checking + heuristique MRV
g3 = copy.deepcopy(grille_initiale)
d3 = init_domaines(g3)
cl3 = [(i, j) for i in range(9) for j in range(9) if g3[i][j] == 0]
c3 = [0]
Forward_Checking_compte(g3, d3, cl3, c3, mrv=True)

print("Nombre de nœuds explorés :")
print("  Backtracking simple      :", c1[0])
print("  Forward Checking         :", c2[0])
print("  Forward Checking + MRV   :", c3[0])
```

On constate que le nombre de nœuds explorés diminue à chaque amélioration de l'algorithme : le Forward Checking permet de détecter les échecs plus tôt grâce à la mise à jour des domaines, et l'heuristique MRV (choisir la variable de plus petit domaine) permet de couper encore plus vite les branches vouées à l'échec, puisqu'on instancie en priorité les cases les plus contraintes (principe de l'échec d'abord vu en cours). Sur cette grille, l'écart reste modeste car elle est déjà relativement facile à résoudre, mais l'effet serait beaucoup plus marqué sur une grille plus difficile ou moins remplie.
